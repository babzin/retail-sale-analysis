```python
df["Purchase Date"] = pd.to_datetime(df["Purchase Date"],format='mixed')
```

```python
df.head()
```

```python
text_cols = [
    "Gender",
    "Product Category",
    "Discount Availed",
    "Location"
]
for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
df["Discount Name"] = df["Discount Name"].fillna("Nothing for you")
```

```python
df.head()
```

```python
print("Duplicate Transaction", df["TID"].duplicated().sum())
```

```python
df["Year"] = df["Purchase Date"].dt.year
df["Month"] = df["Purchase Date"].dt.month_name()
df["Day"] = df["Purchase Date"].dt.day_name()
```

```python
df.head()
```

```python
total_gross = df[' Gross Amount ']
total_net = df[' Net Amount '].sum()
total_discount = df[' Discount Amount (INR) '].sum()
total_transaction = df['TID'].nunique()


print("\n====== KPI SUMMARY =====")
print("Total Gross Revenue: ", total_gross)
print("Total Net Revenue: ", total_net)
print("Total Discount Givem: ", total_discount)
print("Total Transcations: ", total_transaction)
```

```python
df[' Net Amount '].dtype
```

```python
df['Net Amount'] = (
    df[' Net Amount ']
    .replace(r'[\$,]', '', regex=True)
             
            )
```

```python
df.head()
```

```python
df['Net Amount'] = pd.to_numeric(df['Net Amount'], errors = "coerce")
```

```python
df['Net Amount'] = df['Net Amount'].fillna(190.32)
```

```python
df['Net Amount'].sum()
```

```python
df['Gross Amount'] = (
    df[' Gross Amount ']
    .replace(r'[\$,]', '', regex=True)
             
            )
```

```python
df['Gross Amount'] = pd.to_numeric(df['Gross Amount'], errors = "coerce")
```

```python
df['Discount Amount'] = (
    df[' Discount Amount (INR) ']
    .replace(r'[\$,]', '', regex=True)
             
            )
```

```python
df['Discount Amount'] = pd.to_numeric(df['Discount Amount'], errors = "coerce")
```

```python
total_gross = df['Gross Amount'].sum()
total_net = df['Net Amount'].sum()
total_discount = df['Discount Amount'].sum()
total_transaction = df['TID'].nunique()
average_order_value = df['Net Amount'].mean()

print("\n====== KPI SUMMARY =====")
print(f"Total Gross Revenue: {total_gross:,.2f}")
print(f"Total Net Revenue: {total_net:,.2f}")
print(f"Total Discount Given: {total_discount:,.2f}")
print(f"Total Transcations: {total_transaction:,}")
print(f"Average order value: {average_order_value:,.2f}")
```

```python
sns.set_style("whitegrid")
fig, axes = plt.subplots(3, 3, figsize=(22, 16))

fig.suptitle(
    "Retail Sales & Customer Analysis Dashboard",
    fontsize=22,
    fontweight="bold"
)

# 1. SALES BY PRODUCT CATEGORY

category_sales = (
    df.groupby("Product Category")["Net Amount"]
    .sum()
    .sort_values(ascending=False)
)

sns.barplot(
    x=category_sales.values,
    y=category_sales.index,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Net Sales by Product Category")
axes[0, 0].set_xlabel("Net Revenue")
axes[0, 0].set_ylabel("Category")

# 2. SALES BY GENDER

gender_sales = (
    df.groupby("Gender")["Net Amount"]
    .sum()
)

axes[0, 1].pie(
    gender_sales.values,
    labels=gender_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

axes[0, 1].set_title("Revenue Contribution by Gender")

# 3. MONTHLY SALES TREND

monthly_sales = (
    df.groupby(df["Purchase Date"].dt.to_period("M"))["Net Amount"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

sns.lineplot(
    x=monthly_sales.index,
    y=monthly_sales.values,
    marker="o",
    ax=axes[0, 2]
)

axes[0, 2].set_title("Monthly Sales Trend")
axes[0, 2].tick_params(axis='x', rotation=45)

# 4. DISCOUNT USAGE

discount_usage = df["Discount Availed"].value_counts()

sns.barplot(
    x=discount_usage.index,
    y=discount_usage.values,
    ax=axes[1, 0]
)

axes[1, 0].set_title("Discount Usage Count")
axes[1, 0].set_xlabel("Discount Availed")
axes[1, 0].set_ylabel("Count")

# 5. PAYMENT METHOD ANALYSIS

payment_sales = (
    df.groupby("Purchase Method")["Net Amount"]
    .sum()
    .sort_values(ascending=False)
)

sns.barplot(
    x=payment_sales.index,
    y=payment_sales.values,
    ax=axes[1, 1]
)

axes[1, 1].set_title("Revenue by Payment Method")
axes[1, 1].tick_params(axis='x', rotation=30)

# 6. AGE GROUP ANALYSIS

age_sales = (
    df.groupby("Age Group")["Net Amount"]
    .sum()
)

sns.barplot(
    x=age_sales.index,
    y=age_sales.values,
    ax=axes[1, 2]
)

axes[1, 2].set_title("Revenue by Age Group")
axes[1, 2].tick_params(axis='x', rotation=30)

# 7. TOP 10 LOCATIONS

top_locations = (
    df.groupby("Location")["Net Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

sns.barplot(
    x=top_locations.values,
    y=top_locations.index,
    ax=axes[2, 0]
)

axes[2, 0].set_title("Top 10 Cities by Revenue")

# 8. DISTRIBUTION OF NET AMOUNT

sns.histplot(
    df["Net Amount"],
    kde=True,
    bins=30,
    ax=axes[2, 1]
)

axes[2, 1].set_title("Distribution of Net Amount")

# 9. DISCOUNT IMPACT ANALYSIS

sns.boxplot(
    x="Discount Availed",
    y="Net Amount",
    data=df,
    ax=axes[2, 2]
)

axes[2, 2].set_title("Discount Impact on Net Amount")

plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```

```python
top_category = category_sales.idxmax()
print(f"Top Revenue Category: {top_category}")

top_city = top_locations.idxmax()
print(f"Top city: {top_city}")

top_age = age_sales.idxmax()
print(f"the great ones: {top_age}")
```
